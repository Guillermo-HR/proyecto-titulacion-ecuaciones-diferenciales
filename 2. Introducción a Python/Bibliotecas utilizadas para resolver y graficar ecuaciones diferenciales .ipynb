Importar la biblioteca

```python
import numpy as np
```

linspace

```python
valores_x = np.linspace(0, 10, 100)
print(valores_x)
```

meshgrid

```python
valores_x = np.linspace(0, 10, 11)
valores_y = np.linspace(20, 30, 11)
malla = np.meshgrid(valores_x, valores_y, )
print(malla)
```

ones_like

```python
valores_x = np.linspace(0, 10, 100)
vector_unos = np.ones_like(valores_x)
print(vector_unos)
```

asarray

```python
y = [1, 2, 3, 4, 5]
y_arreglo = np.asarray(y, dtype=float)
print(y_arreglo)
```

isinf

```python
y = np.array([1, np.inf, -1])
vector_es_infinito = np.isinf(y)
print(vector_es_infinito)
```

where

```python
arreglo = np.array([1, np.inf, -1])
arreglo = np.where(np.isinf(arreglo), 0, arreglo)
print(arreglo)
```

sqrt

```python
u = np.array([1, 2, 3, 4, 5])
v = np.array([6, 7, 8, 9, 10])
magnitud = np.sqrt(u**2 + v**2)
print(magnitud)
```

arctan2

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([6, 7, 8, 9, 10])
arctan = np.arctan2(y, x)
print(arctan)
```

nanmin

```python
y = np.array([1, 2, 3, 4, 5, np.nan])
minimo = np.nanmin(y)
print(minimo)
```

nanmax

```python
y = np.array([1, 2, 3, 4, 5, np.nan])
maximo = np.nanmax(y)
print(maximo)
```

nan

```python
y = np.array([np.inf])
y_sin_inf = np.where(np.isinf(y), np.nan, y)
print(y_sin_inf)
```

pi

```python
valor_pi = np.pi
print(valor_pi)
```

```python
import sympy as sp
```

```python
from IPython.display import Math
```

Symbol

```python
funcion = sp.E*sp.Derivative(sp.sin(sp.symbols('x')), sp.symbols('x')) + sp.sqrt(sp.symbols('x'))
funcion_latex = sp.latex(funcion)
print(funcion_latex)
display(Math(funcion_latex))
```

```python
x = sp.symbols('x')
u, v = sp.symbols('u v')
display(Math(sp.latex(x+u*v**2)))
```
