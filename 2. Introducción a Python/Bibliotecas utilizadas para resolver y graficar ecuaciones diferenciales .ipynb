latex

```python
expresion = sp.E*sp.Derivative(sp.sin(sp.symbols('x')), sp.symbols('x')) + sp.sqrt(sp.symbols('x'))
expresion_latex = sp.latex(expresion)
print(expresion_latex)
display(Math(expresion_latex))
```

```python
x = sp.symbols('x')
display(Math(sp.latex(x)))
print(type(x))
```

Function

```python
x = sp.symbols('x')
f = sp.Function('f')
display(Math(sp.latex(f)))
print(type(f))
```

```python
x = sp.symbols('x')
f = sp.Function('f')(x)
u, v = sp.symbols('u v')
g = sp.Function('g')(u, v)
display(Math(sp.latex(f)))
display(Math(sp.latex(g)))
```

Derivative

```python
u, v = sp.symbols('u v')
x = sp.symbols('x')
y = sp.Function('y')(x)
f = sp.Function('f')(u, v)
expresion = u+2*v

print("Derivada ordinaria de primer orden de una función")
dydx = sp.Derivative(y, x)
display(Math(sp.latex(dydx)))
print("Derivada ordinaria de segundo orden de una función")
dydxx = sp.Derivative(y, x, 2)
display(Math(sp.latex(dydxx)))
print("Derivada parcial de primer orden de una función")
dfdu = sp.Derivative(f, u)
display(Math(sp.latex(dfdu)))
print("Derivada parcial mixta de una función")
dfd2uv = sp.Derivative(f, u, 2, v)
display(Math(sp.latex(dfd2uv)))
print("Derivada parcial de primer orden de una expresión")
derivada1 = sp.Derivative(expresion, u)
display(Math(sp.latex(derivada1)))
print("Derivada ordinaria de primer orden de una variable")
derivada2 = sp.Derivative(u, v)
display(Math(sp.latex(derivada2)))
```

Eq

```python
u = sp.symbols('u')
b, x, x1, x2, y1, y2 = sp.symbols('b x x_1 x_2 y_1 y_2')
y = sp.Function('y')(x)
f = sp.Function('f')(u)
dfdu = sp.Derivative(f, u)


ecuacion = sp.Eq(u-dfdu, f+2*u)
display(Math(sp.latex(ecuacion)))

m = (y2 - y1)/(x2 - x1)
recta = sp.Eq(y, m*x +b)
display(Math(sp.latex(recta)))
```

lhs

```python
a, b = sp.symbols('a b')
ecuacion = sp.Eq(2*a+3*b, 5-b)
display(Math(sp.latex(ecuacion)))
lado_izquierdo = ecuacion.lhs
display(Math(sp.latex(lado_izquierdo)))
```

rhs

```python
a, b = sp.symbols('a b')
ecuacion = sp.Eq(2*a+3*b, 5-b)
display(Math(sp.latex(ecuacion)))
lado_derecho = ecuacion.rhs
display(Math(sp.latex(lado_derecho)))
```

subs

```python
y, b, x, x1, x2, y1, y2 = sp.symbols('y b x x_1 x_2 y_1 y_2')
m = (y2 - y1)/(x2 - x1)
ecuacion = sp.Eq(y, m*x +b)
display(Math(sp.latex(ecuacion)))
valores = {y1: 2, y2: 5, x1: 1, x2: 3, b: -2}
ecuacion_con_valores = ecuacion.subs(valores)
display(Math(sp.latex(ecuacion_con_valores)))
```

simplify

```python
x = sp.symbols('x')
print("Expresion original")
expresion = (x**3 - x**2 + 3*x + 5)/((x - 1) * (x + 1))
display(Math(sp.latex(expresion)))
expresion_simplificada = sp.simplify(expresion)
print("Expresion simplificada")
display(Math(sp.latex(expresion_simplificada)))
ecuacion = sp.Eq(sp.sin(x)**2 + sp.cos(x)**2, x + 3)
print("Ecuacion original")
display(Math(sp.latex(ecuacion)))
ecuacion_simplificada = sp.simplify(ecuacion)
print("Ecuacion simplificada")
display(Math(sp.latex(ecuacion_simplificada)))
```

solve

```python
x, y = sp.symbols('x y')
ecuacion = sp.Eq(y*x**2+5*y,x)
print("Ecuacion 1")
display(Math(sp.latex(ecuacion)))
solucion_y = sp.solve(ecuacion, y)
print("Lista de soluciones")
print(solucion_y)
print("Primera solución")
display(Math(sp.latex(sp.Eq(y, solucion_y[0]))))
ecuacion_2 = sp.Eq(y**2+5*y,x-1)
print("Ecuacion 2")
display(Math(sp.latex(ecuacion_2)))
solucion_y_2 = sp.solve(ecuacion_2, y)
print("Lista de soluciones")
print(solucion_y_2)
print("Primera solución")
display(Math(sp.latex(sp.Eq(y, solucion_y_2[0]))))
print("Segunda solución")
display(Math(sp.latex(sp.Eq(y, solucion_y_2[1]))))
```

diff

```python
u, v = sp.symbols('u v')
x = sp.symbols('x')
f = sp.Function('y')(x)
g = sp.Function('g')(u,v)
h = sp.Function('h')(u)
expresion1 = sp.sin(x)
expresion2 = x * f
expresion3 = sp.sin(u) * sp.cos(v)
expresion4 = h * g

print(f"Derivada ordinaria de primer orden  d({expresion1})/dx")
derivada1 = sp.diff(expresion1, x)
display(Math(sp.latex(derivada1)))
print(f"Derivada ordinaria de segundo orden d2({expresion1})/dx2")
derivada2 = sp.diff(expresion1, x, 2)
display(Math(sp.latex(derivada2)))
print(f"Derivada ordinaria de primer orden de una expresión con función d({expresion2})/dx")
derivada3 = sp.diff(expresion2, x)
display(Math(sp.latex(derivada3)))
print(f"Derivada parcial de primer orden ∂({expresion3})/∂u")
derivada4 = sp.diff(expresion3, u)
display(Math(sp.latex(derivada4)))
print(f"Derivada parcial mixta ∂3({expresion3})/∂u2∂v")
derivada5 = sp.diff(expresion3, u, 2, v)
display(Math(sp.latex(derivada5)))
print(f"Derivada parcial de primer orden de una expresión con función ∂({expresion4})/∂u")
derivada6 = sp.diff(expresion4, u)
display(Math(sp.latex(derivada6)))
```
