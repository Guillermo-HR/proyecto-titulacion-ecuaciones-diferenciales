```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
R = np.linspace(100, 10000, 100)
C = np.linspace(1e-6, 1000e-6, 100)

R_malla, C_malla = np.meshgrid(R, C)
RC = R_malla * C_malla

fig = plt.figure(figsize=(15, 6))

# Perspectiva enfocada en el cambio de la resistencia R
ax1 = fig.add_subplot(121, projection='3d')
superficie1 = ax1.plot_surface(
    R_malla, C_malla * 1e6, RC,
    cmap='viridis',
    edgecolor='none'
)
ax1.set_xlabel('Resistencia R (Ω)')
ax1.set_ylabel('Capacitancia C (µF)')
ax1.set_zlabel('Constante de tiempo RC (s)')
ax1.set_title('Perspectiva del cambio de R')
ax1.view_init(elev=30, azim=150)

# Perspectiva enfocada en el cambio de la capacitancia C
ax2 = fig.add_subplot(122, projection='3d')
superficie2 = ax2.plot_surface(
    R_malla, C_malla * 1e6, RC,
    cmap='viridis',
    edgecolor='none'
)
ax2.set_xlabel('Resistencia R (Ω)')
ax2.set_ylabel('Capacitancia C (µF)')
ax2.set_zlabel('Constante de tiempo RC (s)')
ax2.set_title('Perspectiva del cambio de C')
ax2.view_init(elev=30, azim=300)

fig.suptitle('Constante de tiempo en un circuito RC', fontsize=14)
fig.colorbar(superficie1, ax=[ax1, ax2], shrink=0.7, label='RC (s)')
plt.show()
```
