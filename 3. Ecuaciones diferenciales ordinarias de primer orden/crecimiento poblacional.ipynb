## Teoría

Modelar el crecimiento poblacional es un problema que tiene aplicaciones en biología, ecología, economía, y otras áreas. Los modelos de crecimiento poblacional permiten predecir cómo cambiará la población de una especie a lo largo del tiempo, considerando factores como la tasa de natalidad, la tasa de mortalidad, capacidad de carga, migración, emigración, entre otros.

Existen varios modelos de crecimiento poblacional:
* Crecimiento exponencial: aumento de la población sin restricciones
* Crecimiento logístico: aumento de la población con un límite de capacidad de carga (es el que se estudiará en este notebook)
* Crecimiento con migración: aumento de la población considerando la entrada de individuos de otras poblaciones
* Crecimiento con emigración: aumento de la población considerando la salida de individuos hacia otras poblaciones

Cualquier modelo de crecimiento poblacional tiene una tasa de crecimiento poblacional, esta tasa es el cambio de la población en un periodo de tiempo. en las ecuaciones se representa como $r$ y se calcula como $r=\frac{dP/dt}{P}=\frac{P_{t+1}-P_t}{P_t}$.

Para mostrar como se puede obtener esta tasa vamos a usar el siguiente ejemplo:
* Un tipo de celula realiza mitosis cada 10 minutos
* La celula tiene una vida de 11 minutos
* Cuando se empieza a estudiar la población de celulas, solo se tiene una celula

En este caso para calcular la tasa de crecimiento poblacional se usa:
* $P_{10}=3$
* $P_{0}=1$
$$r=\frac{P_{10}-P_0}{P_0}=\frac{3-1}{1}=2$$
Pueden existir casos donde la tasa de crecimiento sea negativa, o sea 0; pero el modelo de crecimiento poblacional logístico tradicional solo considera tasas de crecimiento positivas.

Si se utiliza solo la tasa de crecimiento se tiene el modelo exponencial
$$\frac{dP}{dt}=rP$$
Pero ese modelo no es realista en la mayoria de casos ya que no toma en cuenta las limitacines de espacio o recursos del medio ambiente. Para modelar estas limitaciones se tiene la capacidad de carga del ambiente $K$, que es el número máximo de individuos que un ambiente puede sostener. Con la capacidad de carga se agrega un factor de amortiguamiento que reduce la tasa de crecimiento poblacional a medida que la población se acerca a la capacidad de carga y cuando la población es igual a la capacidad de carga, la población deja de crecer. 

Al introducir la capacidad de carga en el modelo exponencial se obtiene el modelo logístico que tiende siempre a la capacidad de carga.
$$\frac{dP}{dt}=rP\left(1-\frac{P}{K}\right)$$
La capacidad de carga solo puede ser un valor positivo ya que representa el número máximo de individuos que un ambiente puede sostener. En algunos libros el modelo se modifica y se representa como:
$$\frac{dP}{dt}=P\left(a-bP\right)$$
Para este cuaderno se va a utilizar la primera forma del modelo logístico, ya que representa explicitamente $r$ y $K$.

Ya que se tiene la ecuación diferencial se puede clasificar, el primer paso para analizarla es extender la ecuación:
$$ \frac{dP}{dt}=rP-\frac{r}{K}P^2$$
* La variable $P$ esta elevada a la segunda potencia, por lo que no es una ED lineal
* Las derivadas solo dependen de 1 variable independiente, por lo que es una ED ordinaria
* El orden de la derivada de mayor orden es 1, por lo que es una ED de primer orden

El modelo logístico es una ecuación diferencial ordinaria de primer orden no lineal.

Ya que se tiene la clasificación se pueden buscar los métodos de solución, en este caso los posibles métodos son variables separables y ecuaciones exactas.

Para validar si puede resolverse por separacion de variables se debe poder expresar como $\frac{dP}{dt}=g(t)h(P)$:
$$\frac{dP}{dt}=(1)\left(rP-\frac{r}{K}P^2\right)$$
$g(t)=1$, $h(P)=rP-\frac{r}{K}P^2$

Si se puede expresar como el producto de una función de $t$ y una función de $P$, por lo que se puede resolver por separacion de variables.

Para validar si puede resolverse por ecuaciones exactas se debe poder expresar como $M(t,P)dt+N(t,P)dP=0$ y cumplir $M_P=N_t$.
$$\left(\frac{r}{K}P^2-rP\right)dt+(1)dP=0$$
$M=\frac{r}{K}P^2-rP$, $N=1$

$M_P=\frac{2r}{K}P-r$, $N_t=0$

Como $M_P \neq N_t$, la ecuación no es exacta.

El último paso antes de empezar a resolver la ecuacion es analizar la ecuacion para entender el comportamiento esperado. Se tienen que identificar los coeficientes y variables y que rangos puede tomar.
* $r$: tasa de crecimiento poblacional, solo puede ser positiva
* $K$: capacidad de carga, solo puede ser positiva
* $P$: población, solo puede ser positiva
* $P(0)$: población inicial, solo puede ser positiva; puede ser mayor, menor o igual a $K$
* $t$: tiempo, solo puede ser positivo

Para entender el comportamiento del modelo vamos a analizar que pasa cuando $t \to \infty$.

Mientras $r$ sea más grande, la población cambiará más rápido; mientras $r$ sea más pequeño, la población cambiará más lentamente. 

Mientras $K$ sea más lejano a $P$ se tardará más tiempo en llegar a la capacidad de carga; mientras $K$ es más cercano a $P$ se tardará menos tiempo en llegar a la capacidad de carga. Cuando $P=K$ el factor de amortiguamiento $\left(1-\frac{P}{K}\right)$ se hace 0 y la población deja de crecer sin importar el resto de parámetros.

El último parametro es $P(0)$, esta tiene 3 posibles casos:
* $P(0)<K$: la población inicial es menor a la capacidad de carga, por lo que la población crecerá hasta llegar a $K$
* $P(0)=K$: la población inicial es igual a la capacidad de carga (población en equilibrio), la solución particular es una constante y la población no crecerá ni decrecerá
* $P(0)>K$: la población inicial es mayor a la capacidad de carga (sobrepoblación), por lo que la población decrecerá hasta llegar a $K$

### Definición de variables simbólicas y la ecuación diferencial

En esta sección se definen las variables simbólicas con SymPy y se define la ecuación diferencial que describe el modelo de crecimiento poblacional con límite de capacidad de carga.

En la teoría ya definimos la ecuación diferencial que describe el crecimiento poblacional con límite de capacidad de carga.
$$\frac{dP}{dt} = rP\left(1-\frac{P}{K}\right)$$
Con esto podemos identificar las variables simbólicas que se necesitan:

|Variable|Descripción|Tipo en Sympy|
|---|---|---|
|$t$|Tiempo|Símbolo|
|$r$|Tasa de crecimiento poblacional|Símbolo|
|$K$|Capacidad de carga|Símbolo|
|$P$|Tamaño de la población en $t$|Función de $t$|
|$\frac{dP}{dt}$|Derivada de $P$ con respecto a $t$|Derivada de $P$ con respecto a $t$|

````python
t = sp.Symbol('t')
r = sp.Symbol('r')
K = sp.Symbol('K')
P = sp.Function('P')(t)
dPdt = sp.Derivative(P, t)
````

Ya que se tienen los simbolos necesarios se puede definir la ecuación diferencial. En este punto los coeficientes y variables no tienen ningun valor asignado

````python
ecuacion_diferencial = sp.Eq(dPdt, r * P * (1 - P / K))
````

### Funciones para analizar, resolver y graficar ecuaciones diferenciales

En esta sección se definen las funciones necesarias para analizar, resolver y graficar ecuaciones diferenciales. Estas funciones se pueden reutilizar para otros problemas de ecuaciones diferenciales.

````python
def analizar_ecuacion_diferencial(variable_dependiente, variable_independiente, ecuacion=None, ecuacion_str=None, M=None, N=None):
    metodo_separacion = False
    metodo_lineal = False
    metodo_exacta = False

    # obtener el símbolo de la variable dependiente para que no sea una función de la variable independiente. Se usa en algunas validaciones
    simbolo_dependente = sp.Symbol(variable_dependiente.func.__name__)
    derivada = sp.Derivative(variable_dependiente, variable_independiente)

    if ecuacion is not None: # Si se proporciona la ecuación diferencial en la forma estándar, intentar obtener M y N a partir de ella
        # Mostrar la ecuación diferencial planteada en la forma estándar
        print("Ecuación diferencial planteada:")
        display(Math(ecuacion_str))

        try:
            expresion = ecuacion.lhs - ecuacion.rhs
            numerador, denominador = sp.fraction(sp.together(expresion))

            N = numerador.coeff(derivada)
            M = numerador.subs(derivada, 0)
        except:
            pass
    
    if ecuacion is None: # Si se proporciona M y N, intentar obtener la ecuación diferencial a partir de ellos
        # Mostrar la ecuación diferencial planteada en la forma Mdx + Ndy = 0
        print("Ecuación diferencial planteada:")
        display(Math(sp.latex(sp.Eq(M * sp.Symbol(f'd{variable_independiente}') + N * sp.Symbol(f'd{simbolo_dependente}'), 0))))
        
        ecuacion = sp.Eq(derivada, -M/N)

    # Validar si se puede resolver con separacion de variables
    if ecuacion.lhs == derivada:
        # Si ya esta en la forma dy/dx=f(x,y) tomar el lado derecho de la ecuacion
        f = ecuacion.rhs
    else:
        # Si no esta en la forma dy/dx=f(x,y) despejar para que del lado izquierdo quede dy/dx
        f = sp.solve(ecuacion, derivada)[0]

    # sustituir la variable dependiente por su símbolo en la expresión f
    expresion = f.subs(variable_dependiente, simbolo_dependente)

    # intentar separar la expresión como g(x)h(y)
    try:
        separacion = sp.separatevars(
            expresion,
            symbols=[
                variable_independiente,
                simbolo_dependente
            ],
            dict=True
        )
        if (variable_independiente in separacion) and (simbolo_dependente in separacion):
            metodo_separacion = True
    except:
        pass
        
    # Validar si se puede resolver con el método lineal
    # Una forma de validar si la ecuacion es lineal es despejar la derivada y obtener la segunda derivada de el otro lado de la ecuacion respecto a la variable dependiente
    # dy/dx=f(x,y) si f_yy=0 entonces es lineal
    f_yy = sp.diff(f, variable_dependiente, 2)

    if sp.simplify(f_yy) == 0:
        metodo_lineal = True

    if M is not None and N is not None: 
        # Validar si se puede resolver con el método exacta
        M_y = sp.diff(M.subs(variable_dependiente, simbolo_dependente), simbolo_dependente)
        N_x = sp.diff(N.subs(variable_dependiente, simbolo_dependente), variable_independiente)

        if sp.simplify(M_y - N_x) == 0:
            metodo_exacta = True

    print("Método separación de variables:", metodo_separacion)
    print("Método lineal:", metodo_lineal)
    print("Método exacta:", metodo_exacta)
````

````python
def obtener_solucion_general(variable_dependiente, variable_independiente, ecuacion=None, M=None, N=None, metodo_definido=None):
    if ecuacion is not None and (M is not None or N is not None):
        # Si se proporciona la ecuación diferencial en la forma estándar, no se deben proporcionar M o N
        return None
    if ecuacion is None and (M is None or N is None):
        # Si no se proporciona la ecuación diferencial en la forma estándar, se deben proporcionar M y N
        return None

    if M is not None and N is not None:
        # Si se pasa M y N validar si es exacta
        simbolo_dependente = sp.Symbol(variable_dependiente.func.__name__)
        # Calcular las derivadas parciales de M y N respecto a la variable dependiente e independiente
        M_y = sp.diff(M.subs(variable_dependiente, simbolo_dependente), simbolo_dependente)
        N_x = sp.diff(N.subs(variable_dependiente, simbolo_dependente), variable_independiente)

        if sp.simplify(M_y - N_x) == 0:
            # Aplicar el metodo de exactas
            g_y = sp.Function('g')(simbolo_dependente)
            # Integrar M respecto a t y agregar la constante g(y)
            f = sp.integrate(M.subs(variable_dependiente, simbolo_dependente), variable_independiente) + g_y
            # Derivada parcial de f respecto ala variable dependiente
            df_dy = sp.diff(f, simbolo_dependente)
            # Igualar la derivada parcial de f respecto a la variable dependiente con N
            df_N = sp.Eq(df_dy, N.subs(variable_dependiente, simbolo_dependente))
            # despejar g'
            dg_dy = sp.solve(df_N, sp.Derivative(g_y, simbolo_dependente))[0]
            # Integrar para obtener g
            g = sp.integrate(dg_dy, simbolo_dependente)
            # Remplazar g en f para obtener la solución general
            f = f.subs(g_y, g)
            solucion_general = sp.Eq(f, sp.Symbol('C')).subs({simbolo_dependente: variable_dependiente})

            return solucion_general
        else:
            ecuacion = sp.Eq(sp.Derivative(variable_dependiente, variable_independiente), -M/N)
    
    # Convertir decimales a racionales para evitar problemas de redondeos o residuos al hacer el cálculo simbólico
    ecuacion = ecuacion.replace(
        lambda x: x.is_Float,
        lambda x: sp.Rational(str(x))
    )

    try:
        if metodo_definido is not None:
            # Si se proporciona un método específico, intentar resolver la ecuación diferencial con ese método
            sol = sp.dsolve(
                ecuacion,
                func=variable_dependiente,
                hint=metodo_definido
            )
        else:
            # Si no se proporciona un método específico, dejar que sympy determine el mejor método automáticamente
            sol = sp.dsolve(
                ecuacion,
                func=variable_dependiente
            )
    except Exception as e:
        print(f"No se pudo resolver la ecuación diferencial: {e}")
        return None
    
    # En algunos casos se devuelve una lista de soluciones, en ese caso se toma la primera solución
    if isinstance(sol, list):
        solucion_general = sol[0]
    else:
        solucion_general = sol

    # Si Sympy puede hacer alguna simplificación de la solución, se hace para que sea más legible
    solucion_general = sp.simplify(solucion_general)
            
    # Redondear los decimales a 2 cifras significativas
    solucion_general = solucion_general.replace(
        lambda x: x.is_Float,
        lambda x: sp.Float(x, 2)
    )
            
    return solucion_general
````

````python
def obtener_solucion_particular(solucion_general, variable_dependiente, variable_independiente, pvi):
    if solucion_general is None:
        # Si no se pasa una solución general, terminar la función
        return None

    solucion_general = solucion_general.replace(
            lambda x: x.is_Float,
            lambda x: sp.Rational(str(x))
        )

    # Buscar el símbolo que se utilizó para representar la constante de integración
    C = next(iter(solucion_general.free_symbols - {variable_dependiente, variable_independiente}))

    # Sustituir las condiciones iniciales en la solución general
    condicion_inicial = solucion_general.subs(pvi)
    # Resolver la ecuacion para obtener el valor de la constante de integración
    soluciones_C = sp.solve(condicion_inicial, C)

    # Validar que se haya encontrado al menos una solución para la constante de integración
    if not soluciones_C:
        display(HTML("<p> Error: No se pudo encontrar una solución para la constante de integración.</p>"))
        return None
    # En algunos casos se devuelve una lista de soluciones, en ese caso se toma la primera solución
    if isinstance(soluciones_C, list):
        valor_C = soluciones_C[0]

    # Sustituir el valor de la constante de integración en la expresión de la solución general
    solucion_particular = solucion_general.subs(C, valor_C)
    # Redondear los decimales a 2 cifras significativas
    solucion_particular = solucion_particular.replace(
        lambda x: x.is_Float,
        lambda x: round(float(x), 2)
    )
    # Simplificar la expresión de la solución particular para que sea más legible
    solucion_particular = sp.simplify(solucion_particular)

    return solucion_particular
````

````python
def graficar_campo_direccional(x_min, x_max, y_min, y_max, variable_dependiente, variable_independiente, ecuacion=None, M=None, N=None):   
    if ecuacion is not None and (M is not None or N is not None):
        # Si se proporciona la ecuación diferencial en la forma estándar, no se deben proporcionar M o N
        return None
    if ecuacion is None and (M is None or N is None):
        # Si no se proporciona la ecuación diferencial en la forma estándar, se deben proporcionar M y N
        return None

    # Definir la derivada de la variable dependiente respecto a la variable independiente
    derivada = sp.Derivative(variable_dependiente, variable_independiente)
    if ecuacion is not None:
        if ecuacion.lhs != derivada:
            # Si la ecuación no está en la forma dy/dx = f(x, y), intentar despejar la derivada
            try:
                ecuacion = sp.Eq(derivada, sp.solve(ecuacion, derivada)[0])
            except Exception as e:
                return None
    if M is not None and N is not None:
        ecuacion = sp.Eq(derivada, -M/N)
     
    # Definir el número de puntos en cada eje para el campo direccional
    n_puntos = 30

    # Convertir la ecuación diferencial a una función que pueda ser evaluada numéricamente
    try:
        f = sp.lambdify((variable_independiente, variable_dependiente), ecuacion.rhs, modules='numpy')
    except Exception as e:
        return None

    # Crear la malla de puntos para el campo direccional
    valores_x = np.linspace(x_min, x_max, n_puntos)
    valores_y = np.linspace(y_min, y_max, n_puntos)
    X, Y = np.meshgrid(valores_x, valores_y)

    # Para graficar el campo direccional, se necesita calcular las componentes U y V del vector. v=(U, V) la pendiente de ese vector es V/U, 
    # la derivada representa la pendiente de la curva en ese punto, por lo que V/U=dy/dx=f(x, y). Para simplificar el calculo, se define U=1 y V=f(x, y)
    U = np.ones_like(X)
    V = f(X, Y)
    # Si algun valor de V es infinito o NaN, se reemplaza por NaN para evitar errores al graficar
    V = np.where(np.isfinite(V), V, np.nan)

    # Se obtiene la magnitud de cada vector, en caso de que la magnitud sea 0, se reemplaza por 1 para evitar división por 0
    magnitud = np.sqrt(U**2 + V**2)
    magnitud = np.where(magnitud == 0, 1, magnitud)

    # Se normalizan las componentes U y V para que todos los vectores tengan la misma longitud
    U_normalizada = U / magnitud
    V_normalizada = V / magnitud

    # Se calcula el ángulo de cada vector y se ajusta para que esté en el rango [0, 2π). El ángulo se utiliza para asignar un color a cada vector
    angulo = np.arctan2(V_normalizada, U_normalizada)
    angulo = (angulo + 2 * np.pi) % (2 * np.pi)

    # Se define la normalización de colores para que el mapa de colores se ajuste al rango de ángulos [0, 2π)
    normalizacion_colores = mcolors.Normalize(vmin=0, vmax=2*np.pi)

    # Crear la gráfica del campo direccional utilizando quiver, que dibuja vectores en cada punto de la malla
    fig, ax = plt.subplots(figsize=(9, 5))
    ax.quiver(X, Y, U_normalizada, V_normalizada, angulo, cmap='RdBu', norm=normalizacion_colores, pivot='mid', scale=50)
    # Dar formato a la gráfica
    ax.set_title("Campo direccional de la ecuación diferencial del crecimiento poblacional")
    ax.set_xlabel(f"{variable_independiente}")
    ax.set_ylabel(f"{variable_dependiente}")
    ax.set_xlim(x_min, x_max)
    ax.set_ylim(y_min, y_max)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['bottom'].set_visible(False)
    ax.spines['left'].set_visible(False)
    plt.close(fig)

    return fig
````

````python
def graficar_solucion_particular(solucion_particular, x_min, x_max, variable_dependiente, variable_independiente):
    if solucion_particular is None:
        return None

    # Definir el número de puntos para graficar la solución particular
    n_puntos = 200

    # Definir el símbolo de la variable dependiente para que no sea una función, eso ayuda a evaluar
    simbolo_dependiente = sp.Symbol(variable_dependiente.func.__name__)
    f = solucion_particular.subs({variable_dependiente: simbolo_dependiente})

    # Crear los valores de x para evaluar la solucion
    valores_x = np.linspace(x_min, x_max, n_puntos)
    # Crear el arreglo para guardar los valores de y
    valores_y = []
    # Evaluar la solución
    for x in valores_x:
        # Sustituir la variable independiente en la función de la solución particular para obtener el valor correspondiente de la variable dependiente
        f_x = f.subs(variable_independiente, x)
        # obtener las posibles soluciones para la variable dependiente
        posibles_y = [y for y in sp.solve(f_x, simbolo_dependiente) if y.is_real]

        # Si hay al menos una solución real, tomar la primera solución y agregarla a la lista de valores de y
        if posibles_y:
            valores_y.append(float(posibles_y[0]))
        else:
            # Si no hay soluciones reales, agregar NaN a la lista de valores de y
            valores_y.append(np.nan)
    valores_y = np.array(valores_y)

    # Si algun valor de y es infinito o NaN, se reemplaza por NaN para evitar errores al graficar
    valores_y = np.where(np.isfinite(valores_y), valores_y, np.nan)

    # Encontrar el valor mínimo y máximo de y
    y_min = np.nanmin(valores_y)
    y_max = np.nanmax(valores_y)
    # Definir los valores que se mostraran en el eje y, se crean 11 valores equidistantes entre el valor mínimo y máximo de y
    # Con esto cada segmento del eje y representa 10% del rango de valores de y
    eje_y = np.linspace(y_min, y_max, 11)

    # Encontrar el valor mínimo y máximo de x
    x_min = np.nanmin(valores_x)
    x_max = np.nanmax(valores_x)
    # Definir los valores que se mostraran en el eje x, se crean 11 valores equidistantes entre el valor mínimo y máximo de x
    # Con esto cada segmento del eje x representa 10% del rango de valores de x
    eje_x = np.linspace(x_min, x_max, 11)

    # Crear la gráfica de la solución particular
    fig, ax = plt.subplots(figsize=(9, 5))
    ax.plot(valores_x, valores_y, color='blue', label='Solución particular')
    # Dar formato a la gráfica 
    ax.set_title("Solución particular de la ecuación diferencial del crecimiento poblacional")
    ax.set_xlabel(f"{variable_independiente}")
    ax.set_ylabel(f"{variable_dependiente}")
    ax.set_xlim(x_min, x_max)
    ax.set_yticks(eje_y)
    ax.set_xticks(eje_x)
    ax.grid(True, which='both', linestyle='--', linewidth=0.5)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['bottom'].set_visible(False)
    ax.spines['left'].set_visible(False)
    plt.close(fig)
    
    return fig
````
