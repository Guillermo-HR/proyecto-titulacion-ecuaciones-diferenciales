## Teoría

Los circuitos RLC en serie son un tipo de circuito que tiene una resistencia, un capacitor y un inductor conectados en serie, existen dos casos de este circuito, el primero es cuando el capacitor se esta cargando (tiene una fuente de voltaje conectada) y el segundo es cuando el capacitor se esta descargando (no tiene fuente de voltaje conectada). En este laboratorio se usa el caso donde se tiene conectada una fuente de voltaje. El capacitor es un elemento que almacena energía en forma de campo eléctrico, mientras que el inductor lo hace en un campo magnético, por lo que ambos son elementos de almacenamiento de energía de primer orden. Al tener 2 elementos de almacenamiento de energía, el circuito RLC es un ejemplo de circuito de segundo orden que se puede modelar con una ecuacion diferencial de segundo orden.

Los elementos de este circuito son una resistencia, un capacitor, un inductor y una fuente de voltaje de corriente directa. El diagrama del circuito RLC en serie es el siguiente:

![Circuito RLC](https://raw.githubusercontent.com/Guillermo-HR/proyecto-titulacion-ecuaciones-diferenciales/771231f2d8df61af98f93811986e1e9509aac2ef/4.%20Ecuaciones%20diferenciales%20lineales%20de%20orden%20superior%20con%20coeficientes%20constantes/circuitoRLC.png)

Diagrama tomado de Dorf, Richar., & Svoboda, James. (2015). Circuitos Eléctricos.

---
Utilizando las leyes de Kirchhoff se tiene que el voltaje de la fuente es igual a la suma del voltaje en la resistencia, el voltaje en el capacitor y el voltaje del inductor, es decir:
$$E = V_R + V_C + V_L$$
Por la ley de Ohm, el voltaje en la resistencia es igual a la corriente por la resistencia $V_R = Ri$, al sustituir esta expresion en la ecuacion de Kirchhoff se tiene:
$$E = Ri + V_C + V_L$$
El voltaje en el inductor se puede expresar como $V_L = L\frac{di}{dt}$, al sustituir esta expresion en la ecuacion anterior se tiene:
$$E = Ri + V_C + L\frac{di}{dt}$$
Como se esta trabajando con una sola malla, la corriente es la misma en todos los elementos del circuito. Se puede expresar la corriente en terminos de la carga en el capacitor con la ecuacion $i = C\frac{dV_C}{dt}$, al sustituir esta expresion en la ecuacion anterior se tiene:
$$E = RC\frac{dV_C}{dt} + V_C + L\frac{d}{dt}\left(C\frac{dV_C}{dt}\right)$$
La capacitancia $C$ es una constante, por lo que se puede sacar de la derivada, ambas derivadas son respecto al tiempo $t$, por lo que se pueden agrupar y reescribir la ecuacion como:
$$E = RC\frac{dV_C}{dt} + V_C + LC\frac{d^2V_C}{dt^2}$$
Por lo general se busca que el coeficiente de la derivada de mayor orden sea 1, al ordenar las derivadas y dividir entre $LC$ se obtiene la ecuacion diferencial para el voltaje en el capacitor:
$$\frac{d^2V_C}{dt^2} + \frac{R}{L}\frac{dV_C}{dt} + \frac{1}{LC}V_C = \frac{E}{LC}$$

---
Ya que se tiene el modelo se puede clasificar la ecuación diferencial 
$$\frac{d^2V_C}{dt^2} + \frac{R}{L}\frac{dV_C}{dt} + \frac{1}{LC}V_C = \frac{E}{LC}$$
* La ecuacion sigue la forma $a_{n}y^{\left(n\right)}+a_{n-1}y^{\left(n-1\right)}+\cdots +a_{1}y^{\prime }+a_{0}y=g\left(x\right)$, por lo que es una ED lineal de coeficientes constantes
* $g(x)$ es diferente de cero, por lo que es una ED no homogénea
* Las derivadas solo dependen de 1 variable independiente, por lo que es una ED ordinaria
* El orden de la derivada de mayor orden es 2, por lo que es una ED de segundo orden

El modelo del circuito RLC en serie con fuente de voltaje es una ecuacion diferencial ordinaria lineal de segundo orden con coeficientes constantes no homogénea.

El último paso antes de empezar a resolver la ecuacion es analizar la ecuacion para entender el comportamiento esperado. Se tienen que identificar los coeficientes y variables y que rangos puede tomar.
* $R$: resistencia, solo puede ser positiva y constante
* $C$: capacitancia, solo puede ser positiva y constante
* $L$: inductancia, solo puede ser positiva y constante
* $E$: voltaje de la fuente, puede ser positiva o negativa dependiendo de la polaridad de la fuente y es constante. En este laboratorio se define que es positivo siguiendo la polaridad de referencia del circuito
* $V_C$: voltaje en el capacitor, puede ser positiva o negativa, es una funcion del tiempo por lo que es variable. En este laboratorio se define que el capacitor es electrolítico y no tiene tolerancia de polaridad, por lo que el voltaje en el capacitor solo puede ser positivo o cero
* $V_C(0)$: voltaje inicial en el capacitor, puede ser positivo, negativo o cero y puede ser mayor, menor o igual al voltaje de la fuente. El caso mas comun es que el capacitor este descargado al inicio, por lo que $V_C(0)=0$. En este laboratorio por la restriccion del capacitor usado solo puede ser positivo o cero
* $i(0)$: corriente inicial en el circuito esta almacenada en el inductor, puede ser positiva, negativa o cero. En este laboratorio se define que es positiva o cero siguiendo la polaridad de referencia del circuito
* $V_C\prime(0)$: derivada del voltaje en el capacitor, puede ser positiva, negativa o cero. Como se tiene $i=CV_C\prime$ y $C$ es positiva, el signo de la derivada del voltaje en el capacitor es igual al signo de la corriente. En este laboratorio por la restriccion del inductor usado solo puede ser positiva o cero
* $t$: tiempo, solo puede ser positivo

### Definición de variables simbólicas y la ecuación diferencial

En esta sección se definen las variables simbólicas con SymPy y se define la ecuación diferencial que describe el circuito RLC.

En la teoría ya definimos la ecuación diferencial que describe la carga de un capacitor en un circuito RLC.
$$\frac{d^2V_C}{dt^2} + \frac{R}{L}\frac{dV_C}{dt} + \frac{1}{LC}V_C = \frac{E}{LC}$$

Con esto podemos identificar las variables simbólicas que se necesitan:

|Variable|Descripción|Tipo en Sympy|
|---|---|---|
|$t$|Tiempo|Símbolo|
|$E$|Voltaje de la fuente|Símbolo|
|$R$|Resistencia|Símbolo|
|$C$|Capacitancia|Símbolo|
|$L$|Inductancia|Símbolo|
|$V_C$|Voltaje del capacitor|Función de $t$|
|$\frac{dV_C}{dt}$|Derivada de $V_C$ con respecto a $t$|Derivada de $V_C$ con respecto a $t$|
|$\frac{d^2V_C}{dt^2}$|Segunda derivada de $V_C$ con respecto a $t$|Segunda derivada de $V_C$ con respecto a $t$|

```python
t = sp.symbols('t')
E = sp.symbols('E')
R = sp.symbols('R')
C = sp.symbols('C')
L = sp.symbols('L')
Vc = sp.Function('Vc')(t)
dVdt = sp.Derivative(Vc, t)
d2Vdt2 = sp.Derivative(Vc, t, t)
```

Ya que se tienen los simbolos necesarios se puede definir la ecuación diferencial. En este punto los coeficientes y variables no tienen ningun valor asignado

```python
ecuacion_diferencial = sp.Eq(d2Vdt2 + (R/L)*dVdt + (1/(L*C))*Vc, E/(L*C))
```

### Funciones para analizar, resolver y graficar ecuaciones diferenciales

En esta sección se definen las funciones necesarias para analizar, resolver y graficar ecuaciones diferenciales. Estas funciones se pueden reutilizar para otros problemas de ecuaciones diferenciales.

```python
def obtener_solucion_general(variable_dependiente, variable_independiente, ecuacion=None):    
    # Convertir decimales a racionales para evitar problemas de redondeos o residuos al hacer el cálculo simbólico
    ecuacion = ecuacion.replace(
        lambda x: x.is_Float,
        lambda x: sp.Rational(str(x))
    )

    try:
        y = sp.dsolve(ecuacion, variable_dependiente)
        # En algunos casos se devuelve una lista de soluciones, en ese caso se toma la primera solución
        if isinstance(y, list):
            y = y[0]
        ecuacion_homogenea = sp.Eq(ecuacion.lhs, 0)
        y_h = sp.dsolve(ecuacion_homogenea, variable_dependiente)
        # En algunos casos se devuelve una lista de soluciones, en ese caso se toma la primera solución
        if isinstance(y_h, list):
            y_h = y_h[0]
        y_p = y.rhs - y_h.rhs
    except Exception as e:
        print(f"No se pudo resolver la ecuación diferencial: {e}")
        return None
            
    return y, y_h, y_p
```

```python
def obtener_solucion_particular(solucion_general, pvi):
    if solucion_general is None:
        # Si no se pasa una solución general, terminar la función
        return None
    # Convertir decimales a racionales para evitar problemas de redondeos o residuos al hacer el cálculo simbólico
    solucion_general = solucion_general.replace(
            lambda x: x.is_Float,
            lambda x: sp.Rational(str(x))
        )

    sistema_ecuaciones = []
    for n, condicion in enumerate(pvi.keys()):
        n_derivada = sp.diff(solucion_general.rhs, t, n)
        ecuacion = sp.Eq(pvi[condicion], n_derivada.subs(t, 0))
        sistema_ecuaciones.append(ecuacion)

    constantes = sp.solve(sistema_ecuaciones, dict=True)

    # Validar que se haya encontrado al menos una solución para la constante de integración
    if not constantes:
        display(HTML("<p> Error: No se pudo encontrar una solución particular.</p>"))
        return None
    
    # Sustituir el valor de la constante de integración en la expresión de la solución general
    solucion_particular = solucion_general.subs(constantes[0])

    return solucion_particular, sistema_ecuaciones
```

```python
def graficar_solucion_particular(solucion_particular, x_min, x_max, variable_dependiente, variable_independiente):
    if solucion_particular is None:
        return None

    # Convertir la solución simbólica en una función numérica
    funcion = sp.lambdify(variable_independiente, solucion_particular.rhs, modules="numpy")

    valores_x = np.linspace(x_min, x_max, 200)

    # Evaluar todos los puntos
    resultado = np.asarray(funcion(valores_x))

    if resultado.ndim == 0:
        valores_y = np.full_like(valores_x, float(resultado))
    else:
        valores_y = np.asarray(np.real_if_close(resultado),dtype=float)

    # Evitar valores infinitos o NaN
    valores_y = np.where(
        np.isfinite(valores_y),
        valores_y,
        np.nan
    )

    # Comprobar que existan valores válidos
    if not np.any(np.isfinite(valores_y)):
        return None

    # Encontrar el valor mínimo y máximo de y
    y_min = np.nanmin(valores_y)
    y_max = np.nanmax(valores_y)
    # Definir los valores que se mostraran en el eje y, se crean 11 valores equidistantes entre el valor mínimo y máximo de y
    # Con esto cada segmento del eje y representa 10% del rango de valores de y
    eje_y = np.linspace(y_min, y_max, 11)

    # Encontrar el valor mínimo y máximo de x
    x_min = np.nanmin(valores_x)
    x_max = np.nanmax(valores_x)
    # Definir los valores que se mostraran en el eje x, se crean 11 valores equidistantes entre el valor mínimo y máximo de x
    # Con esto cada segmento del eje x representa 10% del rango de valores de x
    eje_x = np.linspace(x_min, x_max, 11)

    # Crear la gráfica de la solución particular
    fig, ax = plt.subplots(figsize=(11, 4))
    ax.plot(valores_x, valores_y, color="blue")
    # Dar formato a la gráfica 
    ax.set_title("Solución particular de la ecuación diferencial del circuito RLC")
    ax.set_xlabel(f"{variable_independiente}")
    ax.set_ylabel(f"{variable_dependiente}")
    ax.set_xlim(x_min, x_max)
    ax.set_yticks(eje_y)
    ax.set_xticks(eje_x)
    ax.grid(True, which='both', linestyle='--', linewidth=0.5)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['bottom'].set_visible(False)
    ax.spines['left'].set_visible(False)
    plt.close(fig)

    return fig
```

### Creación de la interfaz

En esta sección se desarrolla la interfaz interactiva junto con todos los componentes necesarios para mostrar los resultados. Se incorporan sliders para modificar los parámetros de la ecuación diferencial y elementos para visualizar los resultados.

```python
# En esta celda se crean los sliders para los parámetros de la ecuación diferencial
# Valores comerciales de resistencias en Ohms
valores_R = [
    100, 120, 150, 180, 220, 270, 330, 390, 470,
    560, 680, 820,
    1000, 1200, 1500, 1800, 2200, 2700, 3300, 3900, 4700,
    5600, 6800, 8200,
    10000
]
# Valores comerciales de capacitores en microfaradios
valores_C = [
    1, 2.2, 3.3, 4.7,
    10, 22, 33, 47,
    100, 220, 330, 470,
    1000
]
# Valores comerciales de inductancias en Henrios
valores_L = [
    0.0001, 0.00022, 0.00047,
    0.001, 0.0022, 0.0047,
    0.01, 0.022, 0.047,
    0.1, 0.22, 0.47,
    1.0
]

# Sliders
slider_R = widgets.SelectionSlider(
    options=valores_R,
    value=1000,
    description='R [Ω]:',
    continuous_update=False
)

slider_C = widgets.SelectionSlider(
    options=valores_C,
    value=100,
    description='C [μF]:',
    continuous_update=False
)

slider_L = widgets.SelectionSlider(
    options=valores_L,
    value=1.0,
    description='L [H]:',
    continuous_update=False
)

slider_E = widgets.IntSlider(
    value=12,
    min=1,
    max=24,
    step=1,
    description='E[V]:',
    continuous_update=False
)

slider_Vc0 = widgets.IntSlider(
    value=0,
    min=0,
    max=30,
    step=1,
    description='Vc(0):',
    continuous_update=False
)

slider_dVcdt_0 = widgets.IntSlider(
    value=0,
    min=0,
    max=30,
    step=1,
    description='Vc\'(0):',
    continuous_update=False
)

slider_t_max = widgets.FloatSlider(
    value=5,
    min=0.01,
    max=60,
    step=0.005,
    description='Rango de tiempo graficado [s]:',
    continuous_update=False,
    style={'description_width': '200px'},
    layout=widgets.Layout(width='450px')
)
```

```python
# En esta celda se crean los componentes para mostrar los resultados
salida_modelo_sin_parametros = widgets.Output()
salida_modelo_con_parametros = widgets.Output()
salida_yh = widgets.Output()
salida_yp = widgets.Output()
salida_solucion_general = widgets.Output()
salida_sistema_ecuaciones = widgets.Output()
salida_solucion_particular = widgets.Output()
salida_grafica_solucion = widgets.Output()
```

```python
# En esta celda se configural los componentes que no cambian
with salida_modelo_sin_parametros:
    display(Math(sp.latex(ecuacion_diferencial)))
```

```python
# En esta celda se define la estructura de la interfaz interactiva, organizando los sliders y las salidas en filas y columnas
controles_parametros = widgets.HBox([
    slider_R,
    slider_C,
    slider_L,
    slider_E
])

controles_condiciones_iniciales = widgets.HBox([
    slider_Vc0,
    slider_dVcdt_0
])

controles_grafica = widgets.HBox([
    slider_t_max
])

controles = widgets.VBox([
    controles_parametros,
    controles_condiciones_iniciales,
    controles_grafica
])

panel_solucion_general = widgets.VBox([
    widgets.HTML("<h3>Ecuación del modelo sin parámetros</h3>"),
    salida_modelo_sin_parametros,
    widgets.HTML("<h3>Ecuación del modelo con parámetros</h3>"),
    salida_modelo_con_parametros,
    widgets.HTML("<h3>Solución general</h3>"),
    salida_solucion_general,
    salida_yh,
    salida_yp
])

panel_solucion_particular = widgets.VBox([
    widgets.HTML("<h3>Solución particular</h3>"),
    salida_solucion_particular,
    widgets.HTML("<h3>Sistema de ecuaciones PVI</h3>"),
    salida_sistema_ecuaciones,
    widgets.HTML("<h3>Gráfica de la solución particular</h3>"),
    salida_grafica_solucion
])

resultados = widgets.GridBox(
    children=[
        panel_solucion_general,
        panel_solucion_particular
    ],
    layout=widgets.Layout(
        width='100%',
        grid_template_columns='32% 67%',
        grid_template_rows='auto',
        grid_gap='1%'
    )
)

interfaz = widgets.VBox([
    widgets.HTML("<h1>Laboratorio de circuito RLC</h1>"),
    controles,
    resultados
])
```

```python
def actualizar_interfaz(change=None): 
    # Obtener los valores de los sliders
    valor_R = slider_R.value
    valor_C = sp.Rational(slider_C.value, 10**6)  # Convertir microfaradios a faradios
    valor_L = sp.Rational(slider_L.value)
    valor_E = slider_E.value
    valor_Vc0 = slider_Vc0.value
    valor_dVcdt_0 = slider_dVcdt_0.value
    valor_t_max = slider_t_max.value 

    # Bloquear los sliders mientras se actualiza la interfaz para evitar cambios durante el cálculo
    slider_R.disabled = True
    slider_C.disabled = True
    slider_L.disabled = True
    slider_E.disabled = True
    slider_Vc0.disabled = True
    slider_dVcdt_0.disabled = True
    slider_t_max.disabled = True

    # Sustituir los valores de los parámetros R, C, L y E en la ecuación diferencial
    ecuacion_con_parametros = ecuacion_diferencial.subs({ 
        R: valor_R, 
        C: valor_C,
        L: valor_L,
        E: valor_E
    }) 

    # Limpiar las salidas de los resultados anteriores
    salida_modelo_con_parametros.clear_output()
    salida_yh.clear_output() 
    salida_yp.clear_output() 
    salida_solucion_general.clear_output() 
    salida_sistema_ecuaciones.clear_output() 
    salida_solucion_particular.clear_output() 
    salida_grafica_solucion.clear_output() 

    # Mostrar la ecuación diferencial con los parámetros sustituidos en la interfaz
    with salida_modelo_con_parametros:
        display(Math(sp.latex(ecuacion_con_parametros)))

    # Obtener la solución general, y_h y_yp de la ecuación diferencial con los parámetros sustituidos y mostrarla
    solucion_general, y_h, y_p  = obtener_solucion_general(variable_dependiente=Vc, variable_independiente=t, ecuacion=ecuacion_con_parametros)
    with salida_solucion_general: 
        display(Math(sp.latex(solucion_general.replace(lambda x: x.is_Float, lambda x: sp.Float(x, 2))))) 
    with salida_yh: 
        display(Math('y_h = ' + sp.latex(y_h.rhs.replace(lambda x: x.is_Float, lambda x: sp.Float(x, 2)))))
    with salida_yp: 
        display(Math('y_p = ' + sp.latex(y_p.replace(lambda x: x.is_Float, lambda x: sp.Float(x, 2)))))

    # Obtener la solución particular de la ecuación diferencial con los parámetros sustituidos y mostrarla
    solucion_particular, sistema_ecuaciones = obtener_solucion_particular(solucion_general=solucion_general, pvi = {Vc.subs(t, 0): valor_Vc0, dVdt.subs(t, 0): valor_dVcdt_0}) 
    with salida_solucion_particular: 
        display(Math(sp.latex(solucion_particular.replace(lambda x: x.is_Float, lambda x: sp.Float(x, 2))))) 
    with salida_sistema_ecuaciones: 
        display(Math(r"\begin{cases}" + r"\\".join(sp.latex(eq) for eq in sistema_ecuaciones) + r"\end{cases}"))

    # Graficar la solución particular de la ecuación diferencial con los parámetros sustituidos y mostrar la gráfica
    fig_solucion = graficar_solucion_particular(solucion_particular=solucion_particular, x_min=0, x_max=valor_t_max, variable_dependiente=Vc, variable_independiente=t) 
    with salida_grafica_solucion: 
        display(fig_solucion) 

    # Desbloquear los sliders después de actualizar la interfaz
    slider_R.disabled = False
    slider_C.disabled = False
    slider_L.disabled = False
    slider_E.disabled = False
    slider_Vc0.disabled = False
    slider_dVcdt_0.disabled = False
    slider_t_max.disabled = False
```

```python
# Configurar los sliders para que llamen a la función actualizar_interfaz cuando su valor cambie
slider_R.observe(actualizar_interfaz, names='value')
slider_C.observe(actualizar_interfaz, names='value')
slider_L.observe(actualizar_interfaz, names='value')
slider_E.observe(actualizar_interfaz, names='value')
slider_Vc0.observe(actualizar_interfaz, names='value')
slider_dVcdt_0.observe(actualizar_interfaz, names='value')
slider_t_max.observe(actualizar_interfaz, names='value')
```

3. Modifica el valor de $V_C(0)$ para que tenga un valor diferente de 0. Que representa tener esas condiciones iniciales? Que pasa con el voltaje del capacitor a medida que pasa el tiempo? 
<details>
<summary>Respuesta</summary>

Tener un capacitor con $V_C(0) \neq 0$ representa que el capacitor estaba conectado a otra fuente de voltaje antes de conectarlo a la fuente actual. Se puede ver como un circuito con un switch que controla que fuente de voltaje esta conectada al capacitor. Si se tiene una fuente $E_1$ y despues de $5\tau$ se cambia a una fuente $E_2$ el capacitor se va a estar cargado al voltaje $E_1$. En este escenario sew tienen 3 posibles casos:
* $E_1 < E_2$: al conectar el capacitor a la fuente $E_2$ el capacitor se va a cargar hasta que su voltaje sea igual al de la fuente $E_2$, se mantiene la relación entre $\tau$ y el tiempo que tarda en cargarse el capacitor
* $E_1 = E_2$: al conectar el capacitor a la fuente $E_2$ el capacitor ya esta cargado al voltaje de la fuente. La solución particular es una constante y el capacitor no se va a cargar ni descargar
* $E_1 > E_2$: al conectar el capacitor a la fuente $E_2$ el capacitor se va a descargar hasta que su voltaje sea igual al de la fuente $E_2$, se mantiene la relación entre $\tau$ y el tiempo que tarda en descargarse el capacitor. En este caso se tarta $\tau$ para que se descargue el capacitor hasta el 63.2% de su voltaje inicial y hasta el 0% de su voltaje inicial.

4. Modifica el valor de $V_C(0)$ para que tenga un valor mayor al de la fuente $E$. Investiga la ecuacion diferencial que describe la descarga de un capacitor en un circuito RC. ¿La ecuación diferencial y su solucion general es la misma que la ecuacion diferencial y solucion general que se muestra en este laboratorio? 
<details>
<summary>Respuesta</summary>

La ecuacion diferencial que describe la descarga de un capacitor en un circuito RC es:
$$\frac{dV_C}{dt} + \frac{1}{RC}V_C = 0$$
La solucion general es:
$$V_C(t) = V_C(0)e^{-\frac{t}{RC}}$$

La diferencia con la ecuacion que se muestra en el laboratorio es que el modelo tradicional descarga el capacitor hasta llegar a $V_C(\infty) = 0$, mientras que el modelo del laboratorio carga el capacitor hasta llegar a $V_C(\infty) = E$. Por lo que la ecuacion diferencial y su solucion general son diferentes. Este tipo de problemas se llaman problemas de valor en la frontera. 

</details>
