## Teoría

Los sistemas resorte/masa amortiguado con movimiento forzado son un tipo de sistema mecánico que tiene una masa sostentida en un resorte, un amortiguador que se opone al movimiento de la masa y una fuerza externa que actua sobre la masa. El resorte es un elemento que almacena energía en forma de energia potencial elastica, mientras que el amortiguador disipa energia en forma de calor por la friccion.

Los elementos de este sistema son una masa, un resorte, un amortiguador y una fuerza externa. En este diagrama al amortiguador es el embolo debajo de la masa, aunque se puede sustituir con cuakquier amortiguador lineal. El diagrama del sistema resorte/masa amortiguado con movimiento forzado es el siguiente:

Al suspender un resorte de un soporte el resorte tiene una longitud $l$ al fijarle una masa $m$ el resorte se elonga una longitud $s$. Por la ley de Hooke, el resorte ejerce una fuerza reparadora en sentido contrario a la elongacion de la forma $F = -ks$, donde $k$ es la constante de resorte. Al punto donde $mg = ks$ se le llama punto de equilibrio. Este punto sera el 0 de referencia del sistema y se define que cualquier posicion debajo de este punto es positiva y cualquier posicion arriba de este punto es negativa. 

El amortiguamiento es un fenomeno que se opone al movimiento de la masa que disipa energia en forma de calor. Un amortiguador lineal es aquel que ejerce una fuerza proporcional a la velocidad de la masa, es decir $F = -\beta v$, donde $\beta$ es la constante de amortiguamiento. el signo negativo se tiene porque el amortiguamiento es opuesto al movimiento de la masa.

Adicionalmente por ser un sistema con movimiento forzado, se tiene una fuerza externa que depende del tiempoque actua sobre la masa. Esta fuerza se define como $f(t)$, esta fuerza puede ser de cualquier tipo. Si se tiene una funcion por tramos se soluciona la ecuacion diferencial por tramos, si se tiene una funcion continua para todo el intervalo se tiene una solucion unica. Para este laboratorio se define que la fuerza externa sigue la forma $f(t)=A\cos(Bt)$.

Ya que se identificaron las fuerzas que actuan sobre la masa, se puede aplicar la segunda ley de Newton para obtener el modelo del sistema. Al plantear la segunda ley de Newton se tiene:
$$F = ma$$
Al sustituir las fuerzas que actuan sobre la masa se tiene:
$$-kx-\beta v(t) + A\cos(Bt) = ma$$
Al sustituir $v(t) = \frac{dx}{dt}$ y $a = \frac{d^2x}{dt^2}$ se tiene:
$$-kx-\beta \frac{dx}{dt} + A\cos(Bt) = m\frac{d^2x}{dt^2}$$
Al ordenar las derivadas se obtiene la siguiente ecuacion:
$$m\frac{d^2x}{dt^2} + \beta \frac{dx}{dt} + kx = A\cos(Bt)$$
Si $\beta = 0$ se tiene un sistema resorte/masa sin amortiguamiento, si $A = 0$ se tiene un sistema resorte/masa amortiguado sin movimiento forzado. Para este laboratorio ambos valores deben ser distintos de cero.

---
Ya que se tiene el modelo se puede clasificar la ecuación diferencial 
$$m\frac{d^2x}{dt^2} + \beta \frac{dx}{dt} + kx = A\cos(Bt)$$
* La ecuacion sigue la forma $a_{n}y^{\left(n\right)}+a_{n-1}y^{\left(n-1\right)}+\cdots +a_{1}y^{\prime }+a_{0}y=g\left(x\right)$, por lo que es una ED lineal de coeficientes constantes
* $g(x)$ es diferente de cero, por lo que es una ED no homogénea
* Las derivadas solo dependen de 1 variable independiente, por lo que es una ED ordinaria
* El orden de la derivada de mayor orden es 2, por lo que es una ED de segundo orden

El modelo del sistema resorte/masa amortiguado con movimiento forzado es una ecuacion diferencial ordinaria lineal de segundo orden con coeficientes constantes no homogénea.

El último paso antes de empezar a resolver la ecuacion es analizar la ecuacion para entender el comportamiento esperado. Se tienen que identificar los coeficientes y variables y que rangos puede tomar.
* $m$: masa, solo puede ser positiva y constante
* $\beta$: coeficiente de amortiguamiento, solo puede ser positivo y constante
* $k$: constante del resorte, solo puede ser positiva y constante
* $A$: amplitud de la fuerza externa, solo puede ser positiva y constante
* $B$: frecuencia de la fuerza externa, solo puede ser positiva y constante
* $x$: posicion de la masa, puede ser positiva, negativa o cero. El sistema de referencia se define en el punto de equilibrio, con el signo positivo hacia abajo y el signo negativo hacia arriba
* $x(0)$: posicion inicial de la masa, puede ser positiva, negativa o cero
* $x\prime(0)$: velocidad inicial de la masa, puede ser positiva, negativa o cero. Si es negativa la masa se mueve hacia arriba, si es positiva la masa se mueve hacia abajo y si es cero la masa esta en reposo (no necesariamente en el punto de equilibrio). El signo de la velocidad inicial no tiene que coincidir con el signo de la posicion inicial
* $t$: tiempo, solo puede ser positivo

### Definición de variables simbólicas y la ecuación diferencial

En esta sección se definen las variables simbólicas con SymPy y se define la ecuación diferencial que describe el sistema resorte/masa amortiguado con movimiento forzado.

En la teoría ya definimos la ecuación diferencial que describe la posicion de la masa es:
$$m\frac{d^2x}{dt^2} + \beta \frac{dx}{dt} + kx = A\cos(Bt)$$

Con esto podemos identificar las variables simbólicas que se necesitan:

|Variable|Descripción|Tipo en Sympy|
|---|---|---|
|$t$|Tiempo|Símbolo|
|$x$|Posición de la masa|Función de $t$|
|$m$|Masa|Símbolo|
|$k$|Constante del resorte|Símbolo|
|$beta$|Constante de amortiguamiento|Símbolo|
|$A$|Amplitud de la fuerza externa|Símbolo|
|$B$|Frecuencia de la fuerza externa|Símbolo|
|$\frac{dx}{dt}$|Derivada de $x$ con respecto a $t$|Derivada de $x$ con respecto a $t$|
|$\frac{d^2x}{dt^2}$|Segunda derivada de $x$ con respecto a $t$|Segunda derivada de $x$ con respecto a $t$|

```python
t = sp.symbols('t')
m = sp.symbols('m')
k = sp.symbols('k')
beta = sp.symbols('beta')
A = sp.symbols('A')
B = sp.symbols('B')
x = sp.Function('x')(t)
dxdt = sp.Derivative(x, t)
d2xdt2 = sp.Derivative(x, t, t)
```

Ya que se tienen los simbolos necesarios se puede definir la ecuación diferencial. En este punto los coeficientes y variables no tienen ningun valor asignado

```python
ecuacion_diferencial = sp.Eq(m*d2xdt2 + beta*dxdt + k*x, A*sp.cos(B*t))
```

### Funciones para analizar, resolver y graficar ecuaciones diferenciales

En esta sección se definen las funciones necesarias para analizar, resolver y graficar ecuaciones diferenciales. Estas funciones se pueden reutilizar para otros problemas de ecuaciones diferenciales.

```python
def obtener_solucion_general(variable_dependiente, ecuacion):    
    # Convertir decimales a racionales para evitar problemas de redondeos o residuos al hacer el cálculo simbólico
    ecuacion = ecuacion.replace(
        lambda x: x.is_Float,
        lambda x: sp.Rational(str(x))
    )

    try:
        y = sp.dsolve(ecuacion, variable_dependiente)
        # En algunos casos se devuelve una lista de soluciones, en ese caso se toma la primera solución
        if isinstance(y, list):
            y = y[0]
        ecuacion_homogenea = sp.Eq(ecuacion.lhs, 0)
        y_h = sp.dsolve(ecuacion_homogenea, variable_dependiente)
        # En algunos casos se devuelve una lista de soluciones, en ese caso se toma la primera solución
        if isinstance(y_h, list):
            y_h = y_h[0]
        y_p = y.rhs - y_h.rhs
    except Exception as e:
        print(f"No se pudo resolver la ecuación diferencial: {e}")
        return None
            
    return y, y_h, y_p
```

```python
def obtener_solucion_particular(solucion_general, pvi):
    if solucion_general is None:
        # Si no se pasa una solución general, terminar la función
        return None
    # Convertir decimales a racionales para evitar problemas de redondeos o residuos al hacer el cálculo simbólico
    solucion_general = solucion_general.replace(
            lambda x: x.is_Float,
            lambda x: sp.Rational(str(x))
        )

    sistema_ecuaciones = []
    for n, condicion in enumerate(pvi.keys()):
        n_derivada = sp.diff(solucion_general.rhs, t, n)
        ecuacion = sp.Eq(pvi[condicion], n_derivada.subs(t, 0))
        sistema_ecuaciones.append(ecuacion)

    constantes = sp.solve(sistema_ecuaciones, dict=True)

    # Validar que se haya encontrado al menos una solución para la constante de integración
    if not constantes:
        display(HTML("<p> Error: No se pudo encontrar una solución particular.</p>"))
        return None
    
    # Sustituir el valor de la constante de integración en la expresión de la solución general
    solucion_particular = solucion_general.subs(constantes[0])

    return solucion_particular, sistema_ecuaciones
```

```python
def graficar_solucion_particular(solucion_particular, x_min, x_max, variable_dependiente, variable_independiente):
    if solucion_particular is None:
        return None

    # Convertir la solución simbólica en una función numérica
    funcion = sp.lambdify(variable_independiente, solucion_particular.rhs, modules="numpy")

    valores_x = np.linspace(x_min, x_max, 2000)

    # Evaluar todos los puntos
    resultado = np.asarray(funcion(valores_x))

    if resultado.ndim == 0:
        valores_y = np.full_like(valores_x, float(resultado))
    else:
        valores_y = np.asarray(np.real_if_close(resultado),dtype=float)

    # Evitar valores infinitos o NaN
    valores_y = np.where(
        np.isfinite(valores_y),
        valores_y,
        np.nan
    )

    # Comprobar que existan valores válidos
    if not np.any(np.isfinite(valores_y)):
        return None

    # Encontrar el valor mínimo y máximo de y
    y_min = np.nanmin(valores_y)
    y_max = np.nanmax(valores_y)
    # Definir los valores que se mostraran en el eje y, se crean 11 valores equidistantes entre el valor mínimo y máximo de y
    # Con esto cada segmento del eje y representa 10% del rango de valores de y
    eje_y = np.linspace(y_min, y_max, 21)

    # Encontrar el valor mínimo y máximo de x
    x_min = np.nanmin(valores_x)
    x_max = np.nanmax(valores_x)
    # Definir los valores que se mostraran en el eje x, se crean 11 valores equidistantes entre el valor mínimo y máximo de x
    # Con esto cada segmento del eje x representa 10% del rango de valores de x
    eje_x = np.linspace(x_min, x_max, 11)

    # Crear la gráfica de la solución particular
    fig, ax = plt.subplots(figsize=(11, 4))
    ax.plot(valores_x, valores_y, color="blue")
    # Dar formato a la gráfica 
    ax.set_title("Solución particular de la ecuación diferencial del\nsistema resorte/masa amortiguado forzado")
    ax.set_xlabel(f"{variable_independiente}")
    ax.set_ylabel(f"{variable_dependiente}")
    ax.set_xlim(x_min, x_max)
    ax.invert_yaxis()
    ax.set_yticks(eje_y)
    ax.set_xticks(eje_x)
    ax.grid(True, which='both', linestyle='--', linewidth=0.5)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['bottom'].set_visible(False)
    ax.spines['left'].set_visible(False)
    plt.close(fig)

    return fig
```

### Creación de la interfaz

En esta sección se desarrolla la interfaz interactiva junto con todos los componentes necesarios para mostrar los resultados. Se incorporan sliders para modificar los parámetros de la ecuación diferencial y elementos para visualizar los resultados.

```python
# En esta celda se crean los sliders para los parámetros de la ecuación diferencial
# Sliders
slider_m = widgets.IntSlider(
    value=25,
    min=1,
    max=50,
    step=1,
    description='m [kg]:',
    continuous_update=False
)

slider_k = widgets.IntSlider(
    value=10,
    min=1,
    max=20,
    step=1,
    description='k [N/m]:',
    continuous_update=False
)

slider_beta = widgets.IntSlider(
    value=25,
    min=1,
    max=50,
    step=1,
    description='β [N·s/m]:',
    continuous_update=False
)

slider_A = widgets.IntSlider(
    value=1,
    min=1,
    max=25,
    step=1,
    description='A [N]:',
    continuous_update=False
)

slider_B = widgets.IntSlider(
    value=1,
    min=1,
    max=10,
    step=1,
    description='B:',
    continuous_update=False
)

slider_x0 = widgets.FloatSlider(
    value=0,
    min=-30,
    max=30,
    step=0.1,
    description='x(0) [m]:',
    continuous_update=False
)

slider_dxdt_0 = widgets.FloatSlider(
    value=0,
    min=-10,
    max=10,
    step=0.1,
    description='x\'(0) [m/s]:',
    continuous_update=False
)

slider_t_max = widgets.IntSlider(
    value=5,
    min=1,
    max=3600,
    step=1,
    description='Rango de tiempo graficado [s]:',
    continuous_update=False,
    style={'description_width': '200px'},
    layout=widgets.Layout(width='450px')
)
```

```python
# En esta celda se crean los componentes para mostrar los resultados
salida_modelo_sin_parametros = widgets.Output()
salida_modelo_con_parametros = widgets.Output()
salida_yh = widgets.Output()
salida_yp = widgets.Output()
salida_solucion_general = widgets.Output()
salida_sistema_ecuaciones = widgets.Output()
salida_solucion_particular = widgets.Output()
salida_grafica_solucion = widgets.Output()
```

```python
# En esta celda se configural los componentes que no cambian
with salida_modelo_sin_parametros:
    display(Math(sp.latex(ecuacion_diferencial)))
```

```python
# En esta celda se define la estructura de la interfaz interactiva, organizando los sliders y las salidas en filas y columnas
controles_parametros = widgets.VBox([
    widgets.HBox([
        slider_m,
        slider_k,
        slider_beta
    ]),
    widgets.HBox([
        slider_A,
        slider_B
    ])
])

controles_condiciones_iniciales = widgets.HBox([
    slider_x0,
    slider_dxdt_0
])

controles_grafica = widgets.HBox([
    slider_t_max
])

controles = widgets.VBox([
    controles_parametros,
    controles_condiciones_iniciales,
    controles_grafica
])

panel_solucion_general = widgets.VBox([
    widgets.HTML("<h3>Ecuación del modelo sin parámetros</h3>"),
    salida_modelo_sin_parametros,
    widgets.HTML("<h3>Ecuación del modelo con parámetros</h3>"),
    salida_modelo_con_parametros,
    widgets.HTML("<h3>Solución general</h3>"),
    salida_solucion_general,
    salida_yh,
    salida_yp
])

panel_solucion_particular = widgets.VBox([
    widgets.HTML("<h3>Solución particular</h3>"),
    salida_solucion_particular,
    widgets.HTML("<h3>Sistema de ecuaciones PVI</h3>"),
    salida_sistema_ecuaciones,
    widgets.HTML("<h3>Gráfica de la solución particular</h3>"),
    salida_grafica_solucion
])

resultados = widgets.GridBox(
    children=[
        panel_solucion_general,
        panel_solucion_particular
    ],
    layout=widgets.Layout(
        width='100%',
        grid_template_columns='32% 67%',
        grid_template_rows='auto',
        grid_gap='1%'
    )
)

interfaz = widgets.VBox([
    widgets.HTML("<h1>Laboratorio de sistema resorte masa amortiguado forzado</h1>"),
    controles,
    resultados
])
```

```python
def actualizar_interfaz(change=None): 
    # Obtener los valores de los sliders
    valor_m = slider_m.value
    valor_k = slider_k.value
    valor_beta = slider_beta.value
    valor_A = slider_A.value
    valor_B = slider_B.value
    valor_x0 = sp.Rational(str(slider_x0.value))
    valor_dxdt_0 = sp.Rational(str(slider_dxdt_0.value))
    valor_t_max = slider_t_max.value 

    # Bloquear los sliders mientras se actualiza la interfaz para evitar cambios durante el cálculo
    slider_m.disabled = True
    slider_k.disabled = True
    slider_beta.disabled = True
    slider_A.disabled = True
    slider_B.disabled = True
    slider_x0.disabled = True
    slider_dxdt_0.disabled = True
    slider_t_max.disabled = True

    # Sustituir los valores de los parámetros R, C, L y E en la ecuación diferencial
    ecuacion_con_parametros = ecuacion_diferencial.subs({ 
        m: valor_m, 
        k: valor_k,
        beta: valor_beta,
        A: valor_A,
        B: valor_B
    }) 

    # Limpiar las salidas de los resultados anteriores
    salida_modelo_con_parametros.clear_output()
    salida_yh.clear_output() 
    salida_yp.clear_output() 
    salida_solucion_general.clear_output() 
    salida_sistema_ecuaciones.clear_output() 
    salida_solucion_particular.clear_output() 
    salida_grafica_solucion.clear_output() 

    # Mostrar la ecuación diferencial con los parámetros sustituidos en la interfaz
    with salida_modelo_con_parametros:
        display(Math(sp.latex(ecuacion_con_parametros)))

    # Obtener la solución general, y_h y_yp de la ecuación diferencial con los parámetros sustituidos y mostrarla
    solucion_general, y_h, y_p  = obtener_solucion_general(variable_dependiente=x, ecuacion=ecuacion_con_parametros)
    with salida_solucion_general: 
        display(Math(sp.latex(solucion_general.replace(lambda x: x.is_Float, lambda x: sp.Float(x, 2))))) 
    with salida_yh: 
        display(Math('x_h = ' + sp.latex(y_h.rhs.replace(lambda x: x.is_Float, lambda x: sp.Float(x, 2)))))
    with salida_yp: 
        display(Math('x_p = ' + sp.latex(y_p.replace(lambda x: x.is_Float, lambda x: sp.Float(x, 2)))))

    # Obtener la solución particular de la ecuación diferencial con los parámetros sustituidos y mostrarla
    solucion_particular, sistema_ecuaciones = obtener_solucion_particular(solucion_general=solucion_general, pvi = {x.subs(t, 0): valor_x0, dxdt.subs(t, 0): valor_dxdt_0}) 
    with salida_solucion_particular: 
        display(Math(sp.latex(solucion_particular.replace(lambda x: x.is_Float, lambda x: sp.Float(x, 2))))) 
    with salida_sistema_ecuaciones: 
        display(Math(r"\begin{cases}" + r"\\".join(sp.latex(eq) for eq in sistema_ecuaciones) + r"\end{cases}"))

    # Graficar la solución particular de la ecuación diferencial con los parámetros sustituidos y mostrar la gráfica
    fig_solucion = graficar_solucion_particular(solucion_particular=solucion_particular, x_min=0, x_max=valor_t_max, variable_dependiente=x, variable_independiente=t) 
    with salida_grafica_solucion: 
        display(fig_solucion) 

    # Desbloquear los sliders después de actualizar la interfaz
    slider_m.disabled = False
    slider_k.disabled = False
    slider_beta.disabled = False
    slider_A.disabled = False
    slider_B.disabled = False
    slider_x0.disabled = False
    slider_dxdt_0.disabled = False
    slider_t_max.disabled = False
```

```python
# Configurar los sliders para que llamen a la función actualizar_interfaz cuando su valor cambie
slider_m.observe(actualizar_interfaz, names='value')
slider_k.observe(actualizar_interfaz, names='value')
slider_beta.observe(actualizar_interfaz, names='value')
slider_A.observe(actualizar_interfaz, names='value')
slider_B.observe(actualizar_interfaz, names='value')
slider_x0.observe(actualizar_interfaz, names='value')
slider_dxdt_0.observe(actualizar_interfaz, names='value')
slider_t_max.observe(actualizar_interfaz, names='value')
```
