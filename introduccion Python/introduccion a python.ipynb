## Comentarios

Existen 2 tipos de comentarios los que son solo de 1 linea y los que son de varias lineas.

Para hacer un comentario de una sola linea se utiliza el simbolo #, mientras que para hacer un comentario de varias lineas se utilizan 3 comillas simples o dobles.

```python
# Esto es un comentario de una sola línea
print("Esta parte si se ejecuta") # Esto es un comentario de una sola línea que no se ejecuta
 
"""
Esto es un comentario de varias líneas
que se puede usar para documentar
el código.
"""
 
'''
Esto también es un comentario de varias líneas
sí dentro se pone una función, se puede usar para documentar y no se ejecuta.
print("Esta parte no se ejecuta")
'''
print("Esta parte si se ejecuta")
```

## Tipos de datos

```python
print(1)
print(type(1))

print(1.5)
print(type(1.5))

print("Hola")
print(type("Hola"))

print(True)
print(type(True))

print([])
print(type([]))

print({"a": 1})
print(type({"a": 1}))

print(())
print(type(()))
```

## Variables

Para asignar una variables solo se tiene que escribir el nombre de la variable seguido del simbolo = y luego el valor que se le quiere asignar a la variable. Python no es un lengueje tipado, por lo que no es necesario declarar el tipo de variable que se esta creando. Despues de crear una variable se puede utilizar en cualquier parte del codigo, modificarla y reasignarle un nuevo valor.

```python
a = 5 # Variable de tipo entero
print(a)

b = 3.14 # Variable de tipo flotante
print(b)

c = a + b # Variable de tipo flotante
print(c)

a = 10.1 # Variable de tipo flotante
c = a + b # Variable de tipo flotante
print(c)

d = "Hola" # Variable de tipo string
print(d)

e = d + " Mundo" # Variable de tipo string
print(e)
```

```python
# definir una lista vacia
l1 = []
print(l1)

# definir una lista con elementos
l2 = [0, 1, 2, 3, 4, 5]
print(l2)
```

```python
print(l2[0]) # imprime el primer elemento de la lista
print(l2[-1]) # imprime el último elemento de la lista
```

```python
# agregar un elemento a la lista usando append()
l1.append("original")
print(l1)

# agregar un elemento a la lista usando insert()
l1.insert(0, "pos 0")
print(l1)
l1.insert(-1, "pos -1")
print(l1)

# agregar un elemento a la lista usando extend()
l1.extend(l2)
print(l1)

# agregar una lista con append()
l1.append(l2)
print(l1)
```

```python
# seleccionar un elemento de la lista con el índice
print(l1[0])
print(l1[-2])

# seleccionar un elemento de una lista dentro de otra lista
print(l1[-1][0])

# seleccionar un elemento con pop
v1 = l1.pop()
print(v1)
print(l1)

v2 = l1.pop(0)
print(v2)
print(l1)
```

```python
print(l2)
print(l2[0:3]) # imprime los elementos desde el índice 0 hasta el índice 2
print(l2[2:-2]) # imprime los elementos desde el índice 2 hasta el índice -3
print(l2[:4]) # imprime los elementos desde el índice 0 hasta el índice 3
print(l2[3:]) # imprime los elementos desde el índice 3 hasta el final de la lista
print(l2[::2]) # imprime los elementos desde el índice 0 hasta el final de la lista con un paso de 2
```
