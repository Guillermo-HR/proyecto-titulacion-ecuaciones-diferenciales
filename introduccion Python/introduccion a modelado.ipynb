```python
import plotly.graph_objects as go
```

```python
def funcion(t_i, t_f, F, C=0):
    v = []
    for t in range(t_i, t_f):
        x_i = (F*t/768) + C
        v.append(x_i)

    return v
```

```python
fig = go.Figure()
t_i = 0
t_f_minutos = 5
t_f = t_f_minutos * 60
t = list(range(t_i, t_f))
F = 10

for c in range(0, 10, 1):
    v = funcion(t_i, t_f, F, C=c)
    fig.add_trace(go.Scatter(x=t, y=v, mode='lines', name=f'v(0)={c}[m/s]'))

fig.update_layout(
    title='Modelo matemático movimiento de un objeto',
    xaxis_title='Tiempo (s)',
    yaxis_title='Velocidad (m/s)',
    showlegend=True
)
fig.show()
```
